# 🧹 Fase 1 y 2: Adquisición de Datos y ETL (CRISP-ML(Q))
Este cuaderno cubre las fases de **Preparación de Datos (ETL)** bajo la metodología CRISP-ML(Q). El objetivo es cargar el dataset sintético de salud mental en España, realizar limpieza, imputación (si es necesario) y transformaciones iniciales, exportando un archivo limpio para el análisis exploratorio y modelado.

```python
import pandas as pd
import numpy as np
import os

print("Pandas versión:", pd.__version__)
print("Numpy versión:", np.__version__)
```

### 1. Carga de datos
Cargamos el archivo `mental_health_spain_final.csv` generado previamente.

```python
data_path = 'mental_health_spain_final.csv'
df = pd.read_csv(data_path)
print(f"Dimensiones originales: {df.shape}")
df.head()
```

### 2. Calidad de Datos (Data Quality Check)
Revisamos tipos de datos, valores nulos y estadísticas descriptivas rápidas.

```python
print("--- Información del Dataset ---")
df.info()

print("\n--- Valores Nulos por Columna ---")
print(df.isnull().sum())
```

### 3. Limpieza y Transformación
Aunque el dataset sintético es limpio, implementamos código defensivo para manejar nulos por si se actualiza con datos reales. Imputaremos numéricos con mediana y categóricos con moda.

```python
# Imputación preventiva de valores faltantes
num_cols = df.select_dtypes(include=[np.number]).columns
cat_cols = df.select_dtypes(exclude=[np.number]).columns

for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Limpieza completada. Valores nulos restantes:", df.isnull().sum().sum())
```

### 4. Guardar Dataset Limpio
Exportamos el dataset procesado como `cleaned_data.csv`.

```python
output_path = 'cleaned_data.csv'
df.to_csv(output_path, index=False, encoding='utf-8')
print(f"Dataset limpio guardado exitosamente en: {output_path}")
```
